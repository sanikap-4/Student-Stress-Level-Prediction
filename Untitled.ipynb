```python
# Remove columns that should not be used as model inputs
model_df = df.drop(
    columns=[
        "Stress_Level",
        "Stress_Level_Num",
        "Mental_Health_Score"
    ]
)

model_df.head()
```

```python
# Create input features (X) and target variable (y)
X = model_df
y = df["Stress_Level"]

print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
# Display the first few input rows and target values
print("Input Features (X):")
display(X.head())

print("\nTarget (y):")
display(y.head())
```

#### Train-Test Split

```python
# Split the dataset into training and testing sets

from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training Features:", X_train.shape)
print("Testing Features:", X_test.shape)
print("Training Target:", y_train.shape)
print("Testing Target:", y_test.shape)
```

```python
# Separate numerical and categorical features

numerical_features = [
    "Age",
    "Avg_Daily_Usage_Hours",
    "Daily_Unlocks",
    "Study_Hours",
    "Physical_Activity_Hours",
    "Sleep_Hours_Per_Night"
]

categorical_features = [
    "Gender",
    "Country",
    "Academic_Level",
    "Most_Used_Platform",
    "Purpose_Of_Use"
]

print("Numerical Features:")
print(numerical_features)

print("\nCategorical Features:")
print(categorical_features)
```

```python
# Import tools required for preprocessing

from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
```

```python
# Create preprocessing steps for numerical and categorical features

preprocessor = ColumnTransformer(
    transformers=[
        (
            "num",
            StandardScaler(),
            numerical_features
        ),
        (
            "cat",
            OneHotEncoder(handle_unknown="ignore"),
            categorical_features
        )
    ]
)

print("Preprocessor created successfully!")
```

```python
OneHotEncoder(handle_unknown="ignore")
```

#### Fit the preprocessor on training data

```python
X_train_processed = preprocessor.fit_transform(X_train)

# Transform the test data using the same preprocessing
X_test_processed = preprocessor.transform(X_test)

print("Training data after preprocessing:", X_train_processed.shape)
print("Testing data after preprocessing:", X_test_processed.shape)
```

```python
preprocessor.fit_transform(X_train)
preprocessor.transform(X_test)
```

### Train Our Random Forest ML Model

```python
# Import the Random Forest classification model

from sklearn.ensemble import RandomForestClassifier
```

```python
# Create the Random Forest model

rf_model = RandomForestClassifier(
    n_estimators=200,
    random_state=42,
    n_jobs=-1
)

print("Random Forest model created successfully!")
```

```python
# Train the Random Forest model on the preprocessed training data

rf_model.fit(X_train_processed, y_train)

print("Random Forest training completed!")
```

```python
# Predict stress levels for the test data

y_pred = rf_model.predict(X_test_processed)

print("Predictions generated successfully!")
print("\nFirst 10 Predictions:")
print(y_pred[:10])
```

```python
# Create a function that predicts stress and returns probabilities

def predict_stress(
    age,
    gender,
    country,
    academic_level,
    platform,
    purpose,
    avg_daily_usage_hours,
    daily_unlocks,
    study_hours,
    physical_activity_hours,
    sleep_hours_per_night
):

    student_data = pd.DataFrame({
        "Age": [age],
        "Gender": [gender],
        "Country": [country],
        "Academic_Level": [academic_level],
        "Most_Used_Platform": [platform],
        "Purpose_Of_Use": [purpose],
        "Avg_Daily_Usage_Hours": [avg_daily_usage_hours],
        "Daily_Unlocks": [daily_unlocks],
        "Study_Hours": [study_hours],
        "Physical_Activity_Hours": [physical_activity_hours],
        "Sleep_Hours_Per_Night": [sleep_hours_per_night]
    })

    prediction = loaded_model.predict(student_data)[0]

    probabilities = loaded_model.predict_proba(student_data)[0]

    classes = loaded_model.classes_

    probability_dict = dict(
        zip(classes, probabilities)
    )

    return prediction, probability_dict
```

```python
# Test the prediction system with a sample student

prediction, probabilities = predict_stress(
    age=21,
    gender="Male",
    country="Other",
    academic_level="Undergraduate",
    platform="Facebook",
    purpose="Networking",
    avg_daily_usage_hours=4.0,
    daily_unlocks=134,
    study_hours=4.5,
    physical_activity_hours=2.2,
    sleep_hours_per_night=6.7
)

print("Predicted Stress Level:", prediction)

print("\nPrediction Probabilities:")

for stress_level, probability in probabilities.items():
    print(f"{stress_level}: {probability * 100:.2f}%")
```

```python
# Create a recommendation engine based on stress level and lifestyle factors

def generate_recommendations(
    stress_level,
    avg_daily_usage_hours,
    daily_unlocks,
    study_hours,
    physical_activity_hours,
    sleep_hours_per_night
):
    
    recommendations = []

    # Stress-level based suggestions
    if stress_level == "Low":
        recommendations.append(
            "Maintain your current healthy routine and keep a consistent sleep schedule."
        )
        recommendations.append(
            "Continue balancing study, social media, physical activity, and personal time."
        )

    elif stress_level == "Medium":
        recommendations.append(
            "Try to maintain a consistent daily routine with planned study and rest periods."
        )
        recommendations.append(
            "Consider taking regular breaks during long study sessions."
        )

    elif stress_level == "High":
        recommendations.append(
            "Consider creating a more structured daily routine with dedicated study and rest periods."
        )
        recommendations.append(
            "Try reducing unnecessary social media use and screen interruptions."
        )
        recommendations.append(
            "Include regular physical activity and adequate sleep in your daily routine."
        )

    elif stress_level == "Very High":
        recommendations.append(
            "Prioritize a consistent sleep and daily routine."
        )
        recommendations.append(
            "Consider reducing unnecessary social media usage and frequent phone checking."
        )
        recommendations.append(
            "Break large academic tasks into smaller, manageable tasks with regular breaks."
        )
        recommendations.append(
            "If stress feels persistent or overwhelming, consider talking to a trusted person, "
            "college counselor, or qualified professional."
        )

    # Lifestyle-specific suggestions
    if sleep_hours_per_night < 6:
        recommendations.append(
            "Your reported sleep duration is relatively low; consider working toward a more consistent sleep routine."
        )

    if avg_daily_usage_hours > 6:
        recommendations.append(
            "Your reported daily social media usage is high; consider setting specific screen-free periods."
        )

    if daily_unlocks > 200:
        recommendations.append(
            "Your phone unlock frequency is high; consider turning off non-essential notifications."
        )

    if study_hours < 2:
        recommendations.append(
            "Consider using short, focused study sessions with planned breaks."
        )

    if physical_activity_hours < 1.5:
        recommendations.append(
            "Consider adding some regular physical activity to your daily routine."
        )

    return recommendations
```

```python
# Test the recommendation engine

recommendations = generate_recommendations(
    stress_level=prediction,
    avg_daily_usage_hours=4.0,
    daily_unlocks=134,
    study_hours=4.5,
    physical_activity_hours=2.2,
    sleep_hours_per_night=6.7
)

print("Personalized Suggestions:\n")

for i, recommendation in enumerate(recommendations, 1):
    print(f"{i}. {recommendation}")
```
